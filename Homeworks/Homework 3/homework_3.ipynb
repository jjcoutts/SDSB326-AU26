# Homework 3: Curating Data
____

## Instructions

The format for this homework is different from previous ones. The goal here is to get you thinking about how you'll construct data for the final analysis. You should approach this problem with an eye toward making your life easier later on: That means writing clear, useful, code that you understand and can reuse later, as well as thinking about how you would construct a dataset that would allow you to do interesting analyses. You don't have to use the results from this project for your final project, but it may provide a useful starting point.

⚠️ **NOTE:** Do not use an API or website we've already used in class. If you're unsure whether your submission meets this expectation, send me an email to ask.

----

**Choose one** of the options below and use your results to create a dataset. Save your results in a `.csv` file and submit this alongside your homework. Your data set should have the following:

* Each cell should contain a single observation. Numbers should be converted to numeric types, and none of the cells should contain nested data.
* Each row should represent a single entity, such as an article, a film, a piece of legislation, a person, etc.

Beyond sharing the code you used to extract your data, include the following in your submission:
* A brief explanation of what your code is doing (with sufficient detail such that someone else could replicate or extend your work).
* A simple demonstration of how someone might use your data OR an explanation of what additional steps you would need to take in order to make your data useful for analysis. 

## Option 1

Visit and extract data from at least 50 unique URLs from a site. Process your results into a .csv file

Your dataset should contain at least three columns with useful information from each page you've scraped. For instance, you might...
- Scrape data from npr.com and include a column for the author, the full text of the article, and the date of publication.
- Explore the list of the top [250 films in IMDB](https://www.imdb.com/chart/top/) and get the director, rating and top billed cast members from each movie.
- Curate data from a pro football statistics page and gather data from player profiles or teams. 



## Option 2

Write code to send at least 50 API queries and process your results into a .csv file. Your dataset should contain at least **three columns** with useful information. For instance, you might...
- Gather data on members of Congress from [the congress.gov API](https://api.congress.gov/) and include a column for the members age, state, and first year in office.
- Use the [FBI API](https://www.fbi.gov/wanted/api) to get the crime, nationality, and reward offered of/for their most wanted fugitives since the 1960s. 
- Try out the [ACLED API](https://apidocs.acleddata.com/) to get characteristics of conflicts around the world.
