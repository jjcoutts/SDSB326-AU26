## Project Guidelines

In your report, you must include:
- An **introduction** describing the motivation for the problem, data, and analysis.
- A clear description of the **data** and the variables you are studying, including how you obtained the data.
- **Descriptive statistics and visualizations** that help the reader understand the data.
- An **analysis section** that looks at relationships between variables from different data sources.
- A **conclusion** that discusses your findings and suggests possible next steps.

Write your report in a Jupyter Notebook, formatting it so that the code is included with the text. Make sure to include a title and section headings by using '#' symbols in Markdown formatting.

Treat this like a business report. You should show a little more code/output than you normally would for that, but the document should be clean, free from spelling/punctuation errors, thoughts written in paragraphs and complete sentences, code should be well comented, and you should avoid lengthy outputs that span thousands of lines. (Comment things out before rendering if you need to.) 

### Introduction (4 pts)

You should start the project with a few paragraphs describing the motivation for your analysis. Set the stage for why this is important, define variables of interest, talk about variables related to your study (e.g., set up the nomological network), establish prior research on your topic, and include your hypotheses (if relevant). Your reader should be engaged, but this should be written scientifically. That is, free from flowery language and objective. You should be citing variable definitions and any non-original information. (APA is preferred, but choose any citation style guide.) 

### Methodology (6 pts)

This section is meant to make your work more reproducibe. You must describe the data collection/curation process and the variables that you created/used in your project. You should provide code here, talk about any joining you did, and make sure everything is transparent and well commented. The output also matters here. For example, do not print out whole lists or DataFrames! (Just pieces of them.) You may also show some basic descriptive statistics to explain the structure of the data. You should be citing all relevant data sources. 

### Visualizations (10 pts)

You must include at least three visualizations as part of your report. This can be included in other sections as approproiate (e.g., in the analysis section). These should be publication-quality, meaning you're making adjustments to the background, colors, etc. to make everything look nice and follow the rules of visualization. Make sure axes are clean, the plot is labeled, you chose an appropriate plot for your data, and so on. 

### Analysis or Final Data Product (10 pts)

You must include some form of **analysis** in your report. If you are combining multiple data sources, this section can be dedicated to looking at the relationships and the additional insights that you can get because of the combined data you created. You may instead have a **data product** with some demonstrated uses. This might be if your project involves a significant amount of work in the data scraping and cleaning steps and is a novel data source (that you have demonstrated does not exist elsewhere). For example, if you really like movies, you may be interested in Rotten Tomatoes, Box Office Mojo, and IMDb. You might combine them for 100s of movies and discuss the implications of having all three sources in one place. (You can't make that your final project, but something similarly complex could be adequate.) Another example might be a process that scrapes webpages for concert times and dates and links them to information about bands. Whatever it is, it should be something that is well-documented and replicable. 

### Conclusion (5 pts)

Write a thoughtful 1-2 paragraph conclusion wrapping up what you did, documenting strengths/limitations, and highlighting next steps/future directions the work could go. End the paper bringing everything together. You should not gloss over the above sections. This is your chance to demonstrate critical thinking and the best skill a data scientist can have is self-critique of their work. 

### References (1 pt)
List of references in chosen style guide. (APA preferred.) 

### Writing/Document Quality (2 pts)
Thoughtful, engaging scientific writing. Minimal typos and appropriate use of subheaders and Markdown. Overall, a clean, professional report. 

### Coding Quality (2 pts)
Code is efficient, readable, optimized, and well-commented.
